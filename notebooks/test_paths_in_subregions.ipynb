````python
# === UTILITY FUNCTIONS FOR WORKING WITH STRUCTURES ===

def get_edge_between_nodes(node1, node2, path_idx):
    """
    Finds the edge connecting two nodes in a specific path
    """
    if path_idx >= len(path_nodes):
        return None

    nodes = path_nodes[path_idx]
    node_edges = path_node_edges[path_idx]

    # Find the position of the nodes in the path
    try:
        pos1 = nodes.index(node1)
        pos2 = nodes.index(node2)

        # Check if they are consecutive nodes
        if abs(pos1 - pos2) == 1:
            edge_pos = min(pos1, pos2)
            if edge_pos < len(node_edges):
                return node_edges[edge_pos]
    except ValueError:
        pass

    return None

def get_nodes_of_edge(edge_idx, path_idx):
    """
    Returns the nodes connected by a specific edge in a path
    """
    if path_idx >= len(path_edge_nodes):
        return None

    edge_nodes = path_edge_nodes[path_idx]
    return edge_nodes.get(edge_idx, None)

def validate_path_consistency(path_idx):
    """
    Validates the consistency between all data structures for a path
    """
    if path_idx >= len(path_nodes):
        return False, "Invalid path index"

    nodes = path_nodes[path_idx]
    edges = path_pairs_of_edges[path_idx] if path_idx < len(path_pairs_of_edges) else []
    node_edges = path_node_edges[path_idx] if path_idx < len(path_node_edges) else []
    edge_nodes = path_edge_nodes[path_idx] if path_idx < len(path_edge_nodes) else {}

    errors = []

    # Check if the number of edges between nodes is correct
    expected_node_edges = len(nodes) - 1 if len(nodes) > 0 else 0
    if len(node_edges) != expected_node_edges:
        errors.append(f"Incorrect number of edges between nodes: {len(node_edges)} vs {expected_node_edges}")

    # Check if all path edges are mapped
    for edge_idx in edges:
        if edge_idx not in edge_nodes:
            errors.append(f"Edge {edge_idx} not mapped in edge_nodes")

    # Check connectivity
    for i in range(len(nodes) - 1):
        if i < len(node_edges):
            edge_idx = node_edges[i]
            if edge_idx != -1 and edge_idx in edge_nodes:
                connected = edge_nodes[edge_idx]
                expected_nodes = {nodes[i], nodes[i+1]}
                actual_nodes = set(connected)
                if expected_nodes != actual_nodes:
                    errors.append(f"Edge {edge_idx} connects {connected} but should connect {list(expected_nodes)}")

    is_valid = len(errors) == 0
    return is_valid, errors

def print_path_summary(path_idx):
    """
    Prints a detailed summary of a specific path
    """
    if path_idx >= len(path_nodes):
        print(f"❌ Path {path_idx} does not exist")
        return

    print(f"📍 PATH SUMMARY {path_idx}:")
    print("-" * 30)

    nodes = path_nodes[path_idx]
    edges = path_pairs_of_edges[path_idx] if path_idx < len(path_pairs_of_edges) else []
    node_edges = path_node_edges[path_idx] if path_idx < len(path_node_edges) else []
    edge_nodes = path_edge_nodes[path_idx] if path_idx < len(path_edge_nodes) else {}

    print(f"Nodes: {nodes}")
    print(f"Edges: {edges}")
    print(f"Connectivity sequence:")

    for i in range(len(nodes) - 1):
        if i < len(node_edges):
            edge_idx = node_edges[i]
            if edge_idx != -1:
                print(f"  {nodes[i]} --[{edge_idx}]--> {nodes[i+1]}")
            else:
                print(f"  {nodes[i]} --[ERROR]--> {nodes[i+1]}")

    # Validation
    is_valid, errors = validate_path_consistency(path_idx)
    if is_valid:
        print("✅ Path valid and consistent")
    else:
        print("❌ Problems found:")
        for error in errors:
            print(f"  • {error}")

# Example usage of the functions
print("🔧 UTILITY FUNCTIONS CREATED:")
print("  • get_edge_between_nodes(node1, node2, path_idx)")
print("  • get_nodes_of_edge(edge_idx, path_idx)")
print("  • validate_path_consistency(path_idx)")
print("  • print_path_summary(path_idx)")
print("\nExample usage:")
print("  print_path_summary(0)  # Analyzes the first path")
print("  validate_path_consistency(0)  # Validates consistency")
````

# 📚 **DOCUMENTAÇÃO DAS ESTRUTURAS DE DADOS**

## 🗄️ **Estruturas Existentes**
- **`path_nodes`**: Lista de listas contendo os nós de cada caminho
- **`path_pairs_of_lines`**: Lista de listas com os pares de linhas de cada caminho  
- **`path_pairs_of_edges`**: Lista de listas com os índices das arestas do grafo que formam cada caminho
- **`all_path_perc`**: Lista com o percentual de cada caminho

## 🆕 **Novas Estruturas Criadas**

### 1. **`path_node_edges`**
- **Tipo**: Lista de listas de inteiros
- **Conteúdo**: Para cada caminho, contém os índices das arestas que conectam nós consecutivos
- **Exemplo**: `[[2, 5, 8], [1, 3]]` significa:
  - Caminho 0: aresta 2 conecta nó[0]→nó[1], aresta 5 conecta nó[1]→nó[2], aresta 8 conecta nó[2]→nó[3]
  - Caminho 1: aresta 1 conecta nó[0]→nó[1], aresta 3 conecta nó[1]→nó[2]

### 2. **`path_edge_nodes`**
- **Tipo**: Lista de dicionários
- **Conteúdo**: Para cada caminho, mapeia cada aresta para os nós que ela conecta
- **Exemplo**: `[{2: [10, 15], 5: [15, 22]}, {1: [8, 12]}]` significa:
  - Caminho 0: aresta 2 conecta nós 10↔15, aresta 5 conecta nós 15↔22
  - Caminho 1: aresta 1 conecta nós 8↔12

## 🔗 **Relações Entre as Estruturas**

```python
# Para um caminho i:

# - Todas as arestas em 'edges' estão em 'edge_nodes'

```
